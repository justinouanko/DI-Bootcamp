## Exercise 1: Understanding Data Visualization

**Why is data visualization important in data analysis?**

When you are working with a dataset of even a few hundred rows, reading through raw numbers to find patterns is slow and error-prone. Visualization compresses that information into a form the human eye can process immediately. You can spot outliers, distributions, correlations and trends in seconds that would take much longer to identify by scanning a table. It also makes it easier to explain findings to others who are not digging into the data themselves.

**What is the purpose of a line graph?**

A line graph is used when data points follow a natural sequence, most often time. The line connecting each point emphasizes direction and continuity, making it easy to see whether something is rising, falling or staying flat over that sequence. It is not the right choice for unordered categories, but as soon as order matters, the line graph is usually the clearest option.

## Exercise 2: Temperature Variation Over a Week

```python
import matplotlib.pyplot as plt

days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
temperatures = [72, 74, 76, 80, 82, 78, 75]

plt.figure(figsize=(9, 5))
plt.plot(days, temperatures, marker="o", color="steelblue", linewidth=2)
plt.xlabel("Day")
plt.ylabel("Temperature (°F)")
plt.title("Temperature Variation Over a Week")
plt.tight_layout()
plt.show()
```

## Exercise 3: Monthly Sales Bar Chart

```python
import matplotlib.pyplot as plt

months = ["January", "February", "March", "April", "May"]
sales = [5000, 5500, 6200, 7000, 7500]

plt.figure(figsize=(9, 5))
plt.bar(months, sales, color="coral")
plt.xlabel("Month")
plt.ylabel("Sales Amount ($)")
plt.title("Monthly Sales for a Retail Store")
plt.tight_layout()
plt.show()
```

## Exercise 4: Distribution of CGPA

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("Student Mental health.csv")

plt.figure(figsize=(9, 5))
sns.histplot(data=df, x="What is your CGPA?", color="mediumpurple", shrink=0.8)
plt.title("Distribution of Students CGPA")
plt.xlabel("CGPA Range")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
```

## Exercise 5: Anxiety Levels Across Genders

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("Student Mental health.csv")

plt.figure(figsize=(9, 5))
sns.barplot(
    data=df,
    x="Choose your gender",
    y="Do you have Anxiety?",
    hue="Do you have Anxiety?",
    palette="Set2",
    estimator=lambda x: (x == "Yes").mean(),
    legend=False,
)
plt.title("Proportion of Students with Anxiety by Gender")
plt.xlabel("Gender")
plt.ylabel("Proportion reporting Anxiety")
plt.tight_layout()
plt.show()
```

## Exercise 6: Age vs Panic Attacks

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("Student Mental health.csv")
df["panic_numeric"] = df["Do you have Panic attack?"].map({"Yes": 1, "No": 0})

plt.figure(figsize=(9, 5))
sns.scatterplot(
    data=df,
    x="Age",
    y="panic_numeric",
    hue="Do you have Panic attack?",
    palette={"Yes": "tomato", "No": "steelblue"},
    s=80,
    alpha=0.7,
)
plt.xlabel("Age")
plt.ylabel("Panic Attack (1 = Yes, 0 = No)")
plt.title("Relationship Between Age and Occurrence of Panic Attacks")
plt.yticks([0, 1], ["No", "Yes"])
plt.tight_layout()
plt.show()
```
